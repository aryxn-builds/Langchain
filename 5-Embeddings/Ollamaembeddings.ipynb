#### Ollama Embeddings (using HuggingFace locally - no server needed)
Ollama requires a local GPU server. Using HuggingFaceEmbeddings as a fully local drop-in replacement.
Models run on CPU via sentence-transformers with no server or API key required.

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
# all-MiniLM-L6-v2: fast lightweight embedding model (384-dim), equivalent to gemma:2b for embeddings
embeddings = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')
```

```python
embeddings
```

```python
r1 = embeddings.embed_documents([
    'Alpha is the first letter of Greek alphabet',
    'Beta is the second letter of Greek alphabet',
])
```

```python
len(r1[0])
```

```python
r1[1]
```

```python
embeddings.embed_query('What is the second letter of Greek alphabet')
```

```python
### Other Embedding Models
### https://www.sbert.net/docs/sentence_transformer/pretrained_models.html
# all-mpnet-base-v2: larger model (768-dim), equivalent to mxbai-embed-large
embeddings2 = HuggingFaceEmbeddings(model_name='sentence-transformers/all-mpnet-base-v2')
text = 'This is a test document.'
query_result = embeddings2.embed_query(text)
query_result
```
